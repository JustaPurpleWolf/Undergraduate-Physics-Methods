## Métodos de Soluciones a EDO

### Método de Euler

Sea $\vec{f}(\vec{x},t):R \subset \mathbb{R}^n \rightarrow \mathbb{R}$ una función y 
$$\frac{d\vec{x}}{dt} = \vec{f}(\vec{x},t)$$
una ecuación diferencial de primer orden. Así, para una malla distribuida uniformemente de separación $\Delta t$, el **método de Euler** establece que,
$$\vec{x}_{n+1} = \vec{x}_{n} + \vec{f}(\vec{x}_{n},t_{n}) \Delta t$$


### Método de Taylor

Sea $\vec{f}(\vec{x},t):[a,b] \rightarrow \mathbb{R}$ una función y
$$ \frac{d\vec{x}}{dt} = \vec{f}(\vec{x},t)$$
una ecuación diferencial de primer orden.
Para una malla distribuida uniformemente, el **método de Taylor** de orden $n$ establece que,
$$ \vec{x}_{n+1} = \vec{x}_{n} + \vec{T}^{(n)}(\vec{x}_n,t_n)$$
donde
$$\vec{T}^{(n)} (x,t)= \delta (\vec{x} \cdot \nabla)\vec{x}_1 + \frac{\delta ^2}{2}\delta (\vec{x} \cdot \nabla)^{2}\vec{x}_2 + ... + \delta (\vec{x} \cdot \nabla)^{n}\vec{x}_n $$
Como sabemos que $\frac{dx}{dt} = x'(t) = f(x,t)$ entonces,
$$\vec{T}^{(n)} (x,t) = \delta \vec{f}(\vec{x},t) + \frac{\delta^2}{2}\delta (\vec{x} \cdot \nabla)\vec{f}(\vec{x},t)+...+\delta (\vec{x} \cdot \nabla)^{n-1}\vec{f}(\vec{x},t)$$
donde 
$$ 
(\vec{x} \cdot \nabla)^{n}\vec{f}(\vec{x},t) = 
\left ( \begin{matrix}
\frac{\partial f_1(\vec{x},t)}{\partial x_2} & \frac{\partial f_1(\vec{x},t)}{\partial x_1} & ... & \frac{\partial f_1(\vec{x},t)}{\partial x_n} & \frac{\partial f_1(\vec{x},t)}{\partial t}\\
\frac{\partial f_2(\vec{x},t)}{\partial x_1} & \frac{\partial f_2(\vec{x},t)}{\partial x_2} & ... & \frac{\partial f_2(\vec{x},t)}{\partial x_n} & \frac{\partial f_2(\vec{x},t)}{\partial t}\\
... & ... & . ..  & ... & ... \\
\frac{\partial f_n(\vec{x},t)}{\partial x_1} & \frac{\partial f_n(\vec{x},t)}{\partial x_2} & ... & \frac{\partial f_n(\vec{x},t)}{\partial x_n} & \frac{\partial f_n(\vec{x},t)}{\partial t}\\
\end{matrix}
\right)
\left ( \begin{matrix}
f_1(\vec{x},t))\\
f_2(\vec{x},t)\\
...\\\
f_n(\vec{x},t)\\
f_t(\vec{x},t)
\end{matrix}
\right)
$$
Así, para el **segundo orden** ($n=2$) tenemos que,
$$ \vec{x}_{n+1} = \delta \vec{f}(\vec{x},t) + \frac{\delta^2}{2}\delta (\vec{x} \cdot \nabla)\vec{f}(\vec{x},t) =\vec{x}_{n} + \delta f(x,t) + \frac{\delta^2}{2}\mathbf{J}(\vec{x},t)\cdot\vec{f}(\vec{x},t)$$
Observemos que para $n=1$ recuperamos el *método de Euler*, es decir el método de Euler es equivalente al método de Taylor de orden 1.

### Método de Runge-Kutta

Sea $\vec{f}(\vec{x},t):[a,b] \rightarrow \mathbb{R}$ una función y
$$ \frac{d\vec{x}}{dt} = \vec{f}(\vec{x},t)$$
una ecuación diferencial de primer orden.
Para una malla distribuida uniformemente, el **método de Runge-Kutta** de segundo orden (RK2) establece que,
$$\vec{x}_{n+1} = \vec{x}_n + \frac{\delta}{2}(\vec{k}_{1,n}+\vec{k}_{2,n})$$
donde,
$$\vec{k}_{1,n} = \vec{f}(\vec{x}_n,t_n)$$
$$\vec{k}_{2,n} = \vec{f}(\vec{x}_n+\delta \vec{k}_{1,n},t_n+ \delta)$$

Para una malla distribuida uniformemente, el **método de Runge-Kutta** de cuarto orden (RK4) establece que,
$$\vec{x}_{n+1} = \vec{x}_n + \frac{\delta}{6}(\vec{k}_{1,n}+2\vec{k}_{2,n}+2\vec{k}_{3,n}+\vec{k}_{4,n})$$
donde,
$$\vec{k}_{1,n} = \vec{f}(\vec{x}_n,t_n)$$
$$\vec{k}_{2,n} = \vec{f}(\vec{x}_n+\frac{\delta}{2}\vec{k}_{1,n},t_n+\frac{\delta}{2})$$
$$\vec{k}_{3,n} = \vec{f}(\vec{x}_n+\frac{\delta}{2}\vec{k}_{2,n},t_n+\frac{\delta}{2})$$
$$\vec{k}_{4,n} = \vec{f}(\vec{x}_n+\delta \vec{k}_{3,n},t_n+\delta)$$

## Funciones

### Soluciones a EDO

```python
class solve:
    def Euler(f,t,x0, **kargs): # f es de la forma f(x,t) 
        h , sol = t[1]-t[0] , np.zeros( (len(x0),len(t)) ) # generar una solución de dimensiones (x,t)
        sol[:,0] = x0 # condicion inicial
        for i in range(len(t)-1):
            sol[:,i+1] = sol[:,i] + h*f(sol[:,i],t[i])  # soluciones para cada entrada de x
        return sol, r'Euler $O(h)$'
        
    def Taylor2(f,t,x0, df = None, **kargs): # f es de la forma f(x,t) 
        h , sol = t[1]-t[0] , np.zeros( (len(x0),len(t)) ) # generar una solución de dimensiones (x,t)
        sol[:,0] = x0 # condicion inicial
        for i in range(len(t)-1):
            sol[:,i+1] = sol[:,i] + h*f(sol[:,i],t[i]) + 0.5*(h**2)*df(sol[:,i],t[i]) # soluciones para x con Taylor a segundo orden
        return sol, r'Taylor $O(h^2)$'
        
    def RK2(f,t,x0, **kargs): # f es de la forma f(x,t)
        h , sol = t[1]-t[0] , np.zeros( (len(x0),len(t)) ) # generar una solución de dimensiones (x,t)
        sol[:,0] = x0 # condicion inicial
        for i in range(len(t)-1):
            k1 = f(sol[:,i],t[i])
            k2 = f(sol[:,i]+h*k1,t[i]+h)
            sol[:,i+1] = sol[:,i] + h*(k1+k2)/2 # soluciones para x con Taylor a segundo orden
        return sol, r'RK2'
        
    def RK4(f,t,x0, **kargs): # f es de la forma f(x,t)
        h , sol = t[1]-t[0] , np.zeros( (len(x0),len(t)) ) # generar una solución de dimensiones (x,t)
        sol[:,0] = x0 # condicion inicial
        for i in range(len(t)-1):
            k1 = f(sol[:,i],t[i])
            k2 = f(sol[:,i]+h*k1/2,t[i]+h/2)
            k3 = f(sol[:,i]+h*k2/2,t[i]+h/2)
            k4 = f(sol[:,i]+h*k3,t[i]+h)
            sol[:,i+1] = sol[:,i] + h*(k1+2*k2+2*k3+k4)/6 # soluciones para x con Taylor a segundo orden
        return sol, r'RK4'
        
    def all(f,t, x0, df= None , **kargs):
        sol_euler, _ = solve.Euler(f,t,x0) # solución numérica con EULER
        sol_taylor2, _ = solve.Taylor2(f,t,x0, df= df) # solución numérica con TAYLOR
        sol_RK2, _ = solve.RK2(f,t,x0)# solución numérica con RK2
        sol_RK4, _ = solve.RK4(f,t,[0,1]) # solución numérica con RK4
        sols, methods = [sol_euler, sol_taylor2, sol_RK2, sol_RK4], [r'Euler $O(h)$', r'Taylor $O(h^2)$',r'RK2', r'RK4'] # soluciones, métodos
        return sols, methods
```

### Graficadores

```python
class plot:
    def solution(f,t, method, h):
        fig = plt.figure(figsize=(12,5))
        plt.suptitle(f'Soluciones a la Ecuación Diferencial' + f' con h = {h}')
        ax = fig.add_subplot(1,2,2)
        bx = fig.add_subplot(1,2,1)
        for i in range(0,len(method)):
            ax.plot(f[i][0,0],f[i][1,0],'x')
            ax.plot(f[i][0,:],f[i][1,:], label = method[i],linewidth = '1')
            ax.set_title(f'Espacio Fase')
            ax.set_xlabel(r'$x$'),ax.set_ylabel(r'$v_{x}$'), ax.legend(loc="upper right")

            bx.plot(t[0],f[i][0,0],'x')
            bx.plot(t,np.array(f[i][0,:]),label = method[i],linewidth = '1')
            bx.set_title(f'Solución')
            bx.set_xlabel(r'$t$'),bx.set_ylabel(r'$x$'), bx.legend(loc="upper right")
        plt.show()
```

## Pruebas

### Oscilador Armónico

Usemos el sistema de segundo orden $\ddot{x} + \omega^2 x = 0$, expresado como sistema de primer orden:$$y_1 = x, \quad y_2 = \dot{x}$$$$\begin{pmatrix} \dot{y}_1 \\ \dot{y}_2 \end{pmatrix} = \begin{pmatrix} y_2 \\ -\omega^2 y_1 \end{pmatrix}$$

```python
def f_oscilador(x, t, omega):
    x1, x2 = x[0], x[1]
    dx1 = x2
    dx2 = -(omega**2) * x1
    return np.array([dx1, dx2])

def df_oscilador(x,t,omega):
    x1, x2 = x[0], x[1]   
    d2x1 = -(omega**2) * x1  # d(dy1)/dt = dy2 = -omega^2 * y1
    d2x2 = -(omega**2) * x2  # d(dy2)/dt = -omega^2 * dy1 = -omega^2 * y2
    return np.array([d2x1, d2x2])
```

```python
#condiciones iniciales
omega= 1.0
x0 = [0,1]
h = 0.1 
tf = 10*np.pi
t = np.arange(0,tf,h) # generación de puntos
```

```python
sol, method = solve.RK2(lambda x,t: f_oscilador(x,t, omega), t, x0)
plot.solution([sol], t, [method], h)
```

```python
sols, methods = solve.all(lambda x,t: f_oscilador(x,t, omega = omega), t, x0, df = lambda x,t: df_oscilador(x,t, omega=omega))
plot.solution(sols, t, methods, h)
```

```python
def accurate_solution(method, f,x0, tf = 10*np.pi,  step = 2,set_acc =1e-4, **kargs):
    real_acc , h = 1 , 0.1 
    t = np.arange(0,tf,h) # generación de puntos
    sol, str_method = method(f,t,x0, **kargs) # solución numérica
    while (real_acc >= set_acc):        
        prev_t = t # asignación de tiempo previo
        prev_sol = sol # asignación de solución previa

        h = h/step # reducir h
        
        t = np.arange(0,tf,h) # generación de puntos
        sol, _ = method(f,t,x0, **kargs) # solución numérica
            
        diff = np.zeros(len(prev_t)) 
        for i in range(0,len(prev_t)): # obtener diferencia entre las soluciones
            diff[i] = abs(prev_sol[0,i]-sol[0,i*step])
        real_acc = np.max(diff)
    print(f'Solución con precisión {set_acc} encontrada en h = {h} para el método {str_method}')
    return sol,t, h

def all_accurate_solution(f,x0, tf = 10*np.pi, step = 2,set_acc =1e-4, **kargs):
    methods = [solve.Euler, solve.Taylor2, solve.RK2, solve.RK4]
    str_methods = [r'Euler $O(h)$', r'Taylor $O(h^2)$',r'RK2', r'RK4']
    sols = []
    T = []
    H =[]
    for method in methods:
        sol ,t, h= solve.accurate(method, lambda x,t: f_oscilador(x,t, omega = omega), x0, tf= tf, df = lambda x,t: df_oscilador(x,t, omega=omega))
        sols.append(sol)
        T.append(t)
        H.append(h)
    return sols, T, str_methods, H
    
def accurate_plot(f,t, method, extra):
    fig = plt.figure(figsize=(12,5))
    plt.suptitle(f'Soluciones a la Ecuación Diferencial' + extra)
    ax = fig.add_subplot(1,2,2)
    bx = fig.add_subplot(1,2,1)
    for i in range(0,len(method)):
        ax.plot(f[i][0,0],f[i][1,0],'x')
        ax.plot(f[i][0,:],f[i][1,:], label = method[i],linewidth = '1')
        ax.set_title(f'Plano Fase')
        ax.set_xlabel(r'$x$'),ax.set_ylabel(r'$v_{x}$'), ax.legend(loc="upper right")

        bx.plot(t[i][0],f[i][0,0],'x')
        bx.plot(t[i],np.array(f[i][0,:]),label = method[i],linewidth = '1')
        bx.set_title(f'Solución')
        bx.set_xlabel(r'$t$'),bx.set_ylabel(r'$x$'), bx.legend(loc="upper right")
        
    plt.show()

solve.accurate = accurate_solution
solve.all_accurate = all_accurate_solution
plot.accurate = accurate_plot
```

```python
sol ,t, h= solve.accurate(solve.RK4, lambda x,t: f_oscilador(x,t, omega = omega), x0, tf = 4*np.pi, df = lambda x,t: df_oscilador(x,t, omega=omega) )
```

```python
sols, t, methods, h = solve.all_accurate(lambda x,t: f_oscilador(x,t, omega = omega), x0, tf = 2*np.pi, df = lambda x,t: df_oscilador(x,t, omega=omega))
```

```python
plot.accurate(sols, t, methods, f' con precisión de {1e-4}')
```
