### Sección 1.
Genera un programa de cómputo (en el lenguaje de tu preferencia) que calcule numéricamente la solución a la ecuación de Laplace en dos dimensiones para un conjunto discreto de puntos, tal y como se muestra en la figura. Los puntos de la frontera exterior se encuentran a un potencial $\phi = 0$, mientras que los puntos de la frontera interna están a un potencial igual a $\phi = 100$.



```python
N = 50                # Número de puntos en cada lado de la malla (N x N)
L = 1.0               # Longitud del dominio de la malla (L x L)
d = L / (N - 1)       # Espaciado de la malla (d)
tolerance = 1e-6      # Tolerancia para criterio de convergencia
max_iteration = 10000 # Límite de iteraciones

phi = np.zeros((N,N)) 
```

```python
# Definición de la frontera 
phi_int = 100

i_min_int = N // 4 
i_max_int = 3 * N // 4 
j_min_int = N // 4 
j_max_int = 3 * N // 4 

phi[j_min_int, i_min_int:i_max_int + 1] = phi_int
phi[j_max_int, i_min_int:i_max_int + 1] = phi_int

phi[j_min_int:j_max_int + 1, i_min_int] = phi_int
phi[j_min_int:j_max_int + 1, i_max_int] = phi_int

# Definición de la frontera exterior
phi_ext = 0

phi[0, :] = phi_ext
phi[N - 1, :] = phi_ext
phi[1:N - 1, 0] = phi_ext
phi[1:N - 1, N - 1] = phi_ext
```

```python
for n in range(max_iteration):
    prev_phi = phi.copy()
    max_diff = 0.0
    # Iterar sobre los puntos (excluyendo las fronteras externas)
    for j in range(1, N - 1):
        for i in range(1, N - 1):
            if not (j >= j_min_int and j <= j_max_int and i >= i_min_int and i <= i_max_int and
                    (j == j_min_int or j == j_max_int or i == i_min_int or i == i_max_int)):  
                
                # Ecuación de Diferencias Finitas
                phi[j, i] = 0.25 * (phi[j + 1, i] + phi[j - 1, i] +
                                    phi[j, i + 1] + phi[j, i - 1])

                diff = abs(phi[j, i] - prev_phi[j, i])
                if diff > max_diff: max_diff = diff

    if max_diff < tolerance:
        print(f"Convergencia alcanzada en {n} iteraciones")
        break
        
    if n == max_iteration - 1:
        print(f"Máximo de {max_iteration} iteraciones alcanzado sin converger.")
```

```python
plt.figure(figsize=(8, 6))
# Mapa de calor
contour = plt.contourf(phi, cmap='viridis', levels=50) 
plt.colorbar(contour, label='Potencial $\phi$')

# Equipotenciales
plt.contour(phi, colors='white', linewidths=0.5, levels=[10, 20, 30, 40, 50, 60, 70, 80, 90])

# Fronteras internas
plt.plot([i_min_int, i_max_int], [j_min_int, j_min_int], 'b-', linewidth=1.5)
plt.plot([i_min_int, i_max_int], [j_max_int, j_max_int], 'b-', linewidth=1.5)
plt.plot([i_min_int, i_min_int], [j_min_int, j_max_int], 'b-', linewidth=1.5)
plt.plot([i_max_int, i_max_int], [j_min_int, j_max_int], 'b-', linewidth=1.5)

# Fronteras externas
plt.plot([0, N-1], [0, 0], 'r-', linewidth=4)
plt.plot([0, N-1], [N-1, N-1], 'r-', linewidth=4)
plt.plot([0, 0], [0, N-1], 'r-', linewidth=4)
plt.plot([N-1, N-1], [0, N-1], 'r-', linewidth=4)

plt.suptitle('Ecuación de Laplace Bidimensional', fontweight = 'bold')
plt.title('Método de relajación')
plt.xlabel('x')
plt.ylabel('y')
plt.gca().set_aspect('equal', adjustable='box')
plt.show()
```

```python
Ex, Ey = np.gradient(phi,L)
Ex, Ey = -Ex,-Ey
E = np.sqrt(Ex**2  + Ey**2)

x, y = np.meshgrid(np.linspace(0, N*L, N), np.linspace(0, N*L, N))

plt.figure(figsize=(8, 6))

# Fronteras internas
plt.plot([i_min_int, i_max_int], [j_min_int, j_min_int], 'b-', linewidth=1.5)
plt.plot([i_min_int, i_max_int], [j_max_int, j_max_int], 'b-', linewidth=1.5)
plt.plot([i_min_int, i_min_int], [j_min_int, j_max_int], 'b-', linewidth=1.5)
plt.plot([i_max_int, i_max_int], [j_min_int, j_max_int], 'b-', linewidth=1.5)

# Fronteras externas
plt.plot([0, N], [0, 0], 'r-', linewidth=1.5)
plt.plot([0, N], [N, N], 'r-', linewidth=1.5)
plt.plot([0, 0], [0, N], 'r-', linewidth=1.5)
plt.plot([N, N], [0, N], 'r-', linewidth=1.5)

# Campo vectorial
plt.quiver(x,y, Ex, Ey, E)

plt.suptitle('Ecuación de Laplace Bidimensional', fontweight = 'bold')
plt.title('Método de relajación')
plt.xlabel('x')
plt.ylabel('y')
plt.gca().set_aspect('equal', adjustable='box')
plt.show()
```

### Sección 2.
Genera un programa para resolver el problema de un círculo en el interior y un cuadrado al exterior, como se muestra en la figura. El cuadrado exterior se encuentra aterrizado (potencial $\phi = 0$), mientras que el círculo interno tiene un potencial $\phi(\theta) = sin^2\theta.$

```python
''' 1. Definición de parámetros '''
R = 20                # Número de puntos en el radio de la malla (R)
Theta = 10                # Número de puntos sobre el ángulo de la malla (Theta)
L = 1.0               # Longitud del dominio de la malla (L x L)
d = L / (N - 1)       # Espaciado de la malla (d)
tolerance = 1e-4      # Tolerancia para criterio de convergencia
max_iteration = 10000 # Límite de iteraciones
```

```python
''' 2. Inicialización de la malla '''
phi = np.zeros((R,Theta))
```

```python
''' 3. Definición de las condiciones de fronteras '''
# Definición de la frontera interior
r_int = R//4
phi_int = np.sin(phi[r_int,:])**2
phi[r_int, :] = phi_int

# Definición de la frontera exterior
phi_ext = 0
phi[R-1,:] = phi_ext
```

```python
'''5. Visualización de Resultados '''
plt.figure(figsize=(8, 6))

# Mapa de calor del potencial
contour = plt.contourf(phi, cmap='viridis', levels=50) 
plt.colorbar(contour, label='Potencial $\phi$')

# Equipotenciales para mejor detalle
plt.contour(phi, colors='white', linewidths=0.5, levels=[10, 20, 30, 40, 50, 60, 70, 80, 90])

# Resaltar las fronteras internas
plt.plot([i_min_int, i_max_int], [j_min_int, j_min_int], 'b-', linewidth=1.5)

# Resaltar las fronteras externas
plt.plot([0, N-1], [0, 0], 'r-', linewidth=3)
plt.plot([0, N-1], [N-1, N-1], 'r-', linewidth=3)
plt.plot([0, 0], [0, N-1], 'r-', linewidth=3)
plt.plot([N-1, N-1], [0, N-1], 'r-', linewidth=3)

plt.suptitle('Ecuación de Laplace Bidimensional', fontweight = 'bold')
plt.title('Método de relajación')
plt.xlabel('x')
plt.ylabel('y')
plt.gca().set_aspect('equal', adjustable='box')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --- 1. Parámetros de la Simulación ---
N = 51          # Número de puntos en el eje x e y (N x N malla)
L = 1.0           # Semilongitud del dominio (el cuadrado va de -L a L)
H = 2 * L         # Longitud total del dominio (2L x 2L)
h = H / (N - 1)   # Espaciado de la malla

# Parámetros del círculo
R = 0.3 * L       # Radio del círculo interno
XC = N // 2       # Coordenada X del centro (índice de malla)
YC = N // 2       # Coordenada Y del centro (índice de malla)

TOLERANCIA = 1e-5 # Criterio de convergencia
MAX_ITER = 20000  # Límite de iteraciones

# --- 2. Inicialización de la Malla ---
# Malla para el potencial
phi = np.zeros((N, N))

# Malla para las coordenadas cartesianas (para calcular r y theta)
x = np.linspace(-L, L, N)
y = np.linspace(-L, L, N)

# --- 3. Aplicación de la Frontera Interna (Círculo) ---

# Recorrer la malla para encontrar y fijar los puntos del círculo
for j in range(1, N - 1):
    for i in range(1, N - 1):
        
        # 3a. Coordenadas relativas al centro del cuadrado/círculo
        dx = x[i]
        dy = y[j]
        
        # 3b. Distancia radial y ángulo polar
        r = np.sqrt(dx**2 + dy**2)
        
        # Condición de Frontera Interna: Si el punto está DENTRO del círculo
        # y cerca del borde (para evitar fijar puntos que no son la frontera).
        # Un enfoque más robusto es fijar todos los puntos que caen en r <= R.
        
        if r <= R:
            # Calcular el ángulo theta (np.arctan2 maneja los cuadrantes correctamente)
            # El eje x positivo tiene theta=0.
            theta = np.arctan2(dy, dx)
            
            # Condición de potencial: phi = sin^2(theta)
            phi[j, i] = np.sin(theta)**2
            
            # NOTA: Para el método de relajación, solo necesitamos fijar los puntos
            # que están *en* la frontera. Si fijamos todos los puntos *dentro* del 
            # círculo, la solución será trivialmente phi = sin^2(theta) en esa región.
            # Para un problema de anillo, se debería usar r = R (o r_min < r < r_max).
            # Para este problema, asumiremos que es una condición de potencial FIJO 
            # aplicado sobre el área del círculo (como un electrodo sólido).
            
            # Alternativa: Si solo quieres fijar los puntos que *definen* el radio R:
            # if abs(r - R) < 0.5 * h: 
            #   phi[j, i] = np.sin(theta)**2

# --- 4. Algoritmo de Relajación de Gauss-Seidel ---
print(f"Iniciando cálculo para una malla {N}x{N}. Radio R={R:.2f}")

for iter_count in range(MAX_ITER):
    max_diff = 0.0
    
    # La matriz 'fixed' nos ayuda a no sobrescribir los puntos de frontera
    # 1. Crear una máscara de puntos fijos
    is_fixed = np.zeros((N, N), dtype=bool)
    
    # Fijos: Frontera externa (bordes de la matriz)
    is_fixed[0, :] = True
    is_fixed[N - 1, :] = True
    is_fixed[:, 0] = True
    is_fixed[:, N - 1] = True
    
    # Fijos: Frontera interna (puntos con r <= R)
    for j in range(N):
        for i in range(N):
            dx = x[i]
            dy = y[j]
            r = np.sqrt(dx**2 + dy**2)
            if r <= R:
                is_fixed[j, i] = True
                
    # Iteración sobre los puntos INTERIORES (que no son fijos)
    for j in range(1, N - 1):
        for i in range(1, N - 1):
            
            if not is_fixed[j, i]:
                phi_old = phi[j, i]
                
                # Ecuación de Diferencias Finitas
                phi[j, i] = 0.25 * (phi[j + 1, i] + phi[j - 1, i] +
                                    phi[j, i + 1] + phi[j, i - 1])
                
                # Criterio de Convergencia
                diff = abs(phi[j, i] - phi_old)
                if diff > max_diff:
                    max_diff = diff

    # Condición de parada
    if max_diff < TOLERANCIA:
        print(f"¡Convergencia alcanzada en {iter_count} iteraciones!")
        break
        
    if iter_count == MAX_ITER - 1:
        print(f"Advertencia: Máximo de {MAX_ITER} iteraciones alcanzado. Máx diff: {max_diff:.2e}")


# --- 5. Visualización de Resultados ---
# Coordenadas para el gráfico
X, Y = np.meshgrid(x, y)

plt.figure(figsize=(9, 7))
# Mapa de calor del potencial
contour = plt.contourf(X, Y, phi, cmap='viridis', levels=50) 
plt.colorbar(contour, label='Potencial $\phi$')

# Contornos para líneas equipotenciales
plt.contour(X, Y, phi, colors='white', linewidths=0.5, levels=np.linspace(0, 1, 10))

# Resaltar la frontera interna
circle = plt.Circle((0, 0), R, color='blue', fill=False, linewidth=2)
plt.gca().add_artist(circle)
plt.title(r'Solución a $\nabla^2\phi = 0$ con $\phi_{int} = \sin^2 \theta$')
plt.xlabel(r'$\hat{e}_x$')
plt.ylabel(r'$\hat{e}_y$')
plt.gca().set_aspect('equal', adjustable='box')
plt.xlim(-L, L)
plt.ylim(-L, L)
plt.show()
```

```python
# Usando h del problema (ejemplo: h = 2*L / (N - 1))
L = 1.0 # Semilongitud del dominio
N = phi.shape[0] # Tamaño de la malla
H = 2 * L
h = H / (N - 1)

E_y_raw, E_x_raw = np.gradient(phi, h) 
E_x = -E_x_raw
E_y = -E_y_raw

# Calcular la magnitud del campo (opcional, para visualización)
E_mag = np.sqrt(E_x**2 + E_y**2)

# --- 2. Preparar el Gráfico de Flechas (Quiver Plot) ---

# Submuestreo para mejor visualización (ejemplo: mostrar una flecha cada 5 puntos)
step = 5 
X, Y = np.meshgrid(np.linspace(-L, L, N), np.linspace(-L, L, N))

# Crea una nueva figura para el campo E
plt.figure(figsize=(9, 7))

# 2a. Gráfico del Campo de Potencial (Mapa de Calor)
# Usar la magnitud del campo para el mapa de color
plt.pcolormesh(X, Y, E_mag, cmap='jet')
plt.colorbar(label='Magnitud del Campo Eléctrico |E|')

# 2b. Gráfico de Flechas
# Plotea el submuestreo de las componentes E_x y E_y
plt.quiver(X[::step, ::step], 
           Y[::step, ::step], 
           E_x[::step, ::step], 
           E_y[::step, ::step], 
           scale=np.max(E_mag) * (step * 2), # Ajusta la escala de las flechas
           color='white', 
           headlength=4, 
           headwidth=3) 

plt.title('Campo Eléctrico (Flechas) y Magnitud de E (Color)')
plt.xlabel(r'$\hat{e}_x$')
plt.ylabel(r'$\hat{e}_y$')
plt.gca().set_aspect('equal', adjustable='box')
plt.show()
```
