```python
import numpy as np
import matplotlib.pyplot as plt
import os
from matplotlib.ticker import MultipleLocator, FuncFormatter
```

```python
def pi_formatter(val, pos):
    if val == 0:
        return "0"
    elif val == np.pi:
        return r"$\pi$"
    elif val == -np.pi:
        return r"$-\pi$"
    elif val == np.pi / 2:
        return r"$\pi/2$"
    elif val == -np.pi / 2:
        return r"$-\pi/2$"
    elif val == np.pi / 4:
        return r"$\pi/4$"
    elif val == -np.pi / 4:
        return r"$-\pi/4$"
    elif val == np.pi / 8:
        return r"$\pi/8$"
    elif val == -np.pi / 8:
        return r"$-\pi/8$"
    elif val == np.pi / 16:
        return r"$\pi/16$"
    elif val == -np.pi / 16:
        return r"$-\pi/16$"
    elif val == 2 * np.pi:
        return r"$2\pi$"
    elif val == -2 * np.pi:
        return r"$-2\pi$"
    # Handle other fractions of pi
    else:
        # You can add more complex logic here to represent other fractions
        # For example, to represent 3pi/2 as r"$3\pi/2$"
        fraction = val / np.pi
        if fraction == int(fraction):
            return r"$%d\pi$" % int(fraction)
        else:
            return r"$%s\pi$" % str(fraction)
```

```python
fig = plt.figure(figsize=(8,8))
ax = fig.add_subplot(2,1,1)
bx = fig.add_subplot(2,1,2)

L = 1
dots = 1000
v = np.linspace(-7*L, 7*L, dots)
t = np.linspace(-2*L, 2*L, dots)

F = (L*np.sinc(L*v))**2
f = np.zeros(dots)

for i in range(dots):
    if np.abs(t[i]) < L: 
        f[i] = L-np.abs(t[i])
        
ax.plot(t, f, linewidth = 1, label = f"$f(t)$", color = 'b')
bx.plot(v, F, linewidth = 1, label = f"$F(v)$", color = 'r')

ax.set_title(f"$f(t) = L-|t|$ y su transformada de Fourier para $L=1$")
ax.legend(), bx.legend()
ax.set_xlabel("Espacio del dominio [x]"), bx.set_xlabel("Espacio de frecuencias [k]")
ax.set_ylabel("Espacio de la imágen [f]"), bx.set_ylabel("Espacio de la imágen [Re(F)]")
ax.grid(), bx.grid()
fig.tight_layout()
plt.savefig("fourier-transform-1.png")
plt.show()
```

```python
fig = plt.figure(figsize=(8,8))
ax = fig.add_subplot(2,1,1)
bx = fig.add_subplot(2,1,2)

l = 1
v0 = 1
dots = 1000
v = np.linspace(-10*l, 10*l, dots)
t = np.linspace(-2*l, 2*l, dots)

F = 0.5*l*(np.sinc(l*(v0-v))+np.sinc(l*(v0+v)))
f = np.zeros(dots)

for i in range(dots):
    if np.abs(t[i]) < 0.5*l: 
        f[i] = np.sin(2*np.pi*v0*t[i])
        
ax.plot(t, f, linewidth = 1, label = f"$f(t)$", color ='b')
bx.plot(v, F, linewidth = 1, label = f"$F(v)$", color = 'r')

ax.set_title(f"$f(x) = sin(k_0x)$ y su transformada de Fourier para $\lambda=1$ y $k_0 = 2 \pi$")
ax.legend(), bx.legend()
ax.set_xlabel("Espacio del dominio [x]"), bx.set_xlabel("Espacio de frecuencias [k]")
ax.set_ylabel("Espacio de la imágen [f]"), bx.set_ylabel("Espacio de la imágen [Im(F)]")
ax.grid(), bx.grid()
fig.tight_layout()
plt.savefig("fourier-transform-2.png")
plt.show()
```

```python
fig = plt.figure(figsize=(8,8))
ax = fig.add_subplot(2,1,1)
bx = fig.add_subplot(2,1,2)

l = 1
v0 = 1
dots = 1000
v = np.linspace(-10*l, 10*l, dots)
t = np.linspace(-l, l, dots)

F = 0.25*l*(2*np.sinc(l*v)+np.sinc(l*(2*v0-v))+np.sinc(l*(2*v0+v)))
f = np.zeros(dots)

for i in range(dots):
    if np.abs(t[i]) < 0.5*l: 
        f[i] = (np.cos(2*np.pi*v0*t[i]))**2
        
ax.plot(t, f, linewidth = 1, label = f"$f(t)$", color = 'b')
bx.plot(v, F, linewidth = 1, label = f"$F(v)$", color = 'r')

plt.suptitle(f"$f(t) = cos^2(w_0t)$ y su transformada de Fourier para $t_0=1$ y $w_0=2\pi$")
ax.legend(), bx.legend()
ax.set_xlabel("Espacio del dominio [x]"), bx.set_xlabel("Espacio de frecuencias [k]")
ax.set_ylabel("Espacio de la imágen [f]"), bx.set_ylabel("Espacio de la imágen [Re(F)]")
ax.grid(), bx.grid()
fig.tight_layout()
plt.savefig("fourier-transform-3.png")
plt.show()
```
