```python
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
fig = plt.figure(figsize=(8,4))
ax = fig.add_subplot(1,1,1)
N = [2,6,8,10,16]
t = np.linspace(-2, 2, 1000)
ax.plot(t, np.abs(t), linewidth = 1,linestyle = "--" , label = f"$f(t) = |t|$")
for n in N:
    F = 1
    for i in range(1,n+1):
        f = (4/(np.pi**2))*((-1)**i-1)/(i**2)*np.cos(np.pi*i*t/2)
        F = F + f
    ax.plot(t, F, linewidth = 1, label = f"$f_{{n = {n}}}(t)$")
ax.set_title("f(t) = |t| y su serie de Fourier para distintos valores de n")
ax.legend()
ax.grid()
fig.tight_layout()
plt.savefig("fourier-series-1.png")
plt.show()
```
