```python
import numpy as np
import matplotlib.pyplot as plt
import os
from scipy.signal import unit_impulse
from matplotlib.ticker import MultipleLocator, FuncFormatter
```

```python
def pi_formatter(val, pos):
    if val == 0:
        return "0"
    elif val == np.pi:
        return r"$\pi$"
    elif val == -np.pi:
        return r"$-\pi$"
    elif val == np.pi / 2:
        return r"$\pi/2$"
    elif val == -np.pi / 2:
        return r"$-\pi/2$"
    elif val == np.pi / 4:
        return r"$\pi/4$"
    elif val == -np.pi / 4:
        return r"$-\pi/4$"
    elif val == np.pi / 8:
        return r"$\pi/8$"
    elif val == -np.pi / 8:
        return r"$-\pi/8$"
    elif val == np.pi / 16:
        return r"$\pi/16$"
    elif val == -np.pi / 16:
        return r"$-\pi/16$"
    elif val == 2 * np.pi:
        return r"$2\pi$"
    elif val == -2 * np.pi:
        return r"$-2\pi$"
    # Handle other fractions of pi
    else:
        # You can add more complex logic here to represent other fractions
        # For example, to represent 3pi/2 as r"$3\pi/2$"
        fraction = val / np.pi
        if fraction == int(fraction):
            return r"$%d\pi$" % int(fraction)
        else:
            return r"$%s\pi$" % str(fraction)
```

```python
N = 5
specified_points = [1, N]
x_min, x_max = 0, 7

title = f"Espectros de Amplitud y Fase para $\psi (x,0)$, con $A_{{01}} = 1, N={N}$"

x = np.arange(x_min, x_max + 1)
amplitude = np.zeros_like(x, dtype=int)
phase = np.zeros_like(x, dtype=float)
for point in specified_points:
    amplitude[x == point] = 1
    phase[x==point] = -np.pi/2
    
fig = plt.figure(figsize=(8,8))
ax = fig.add_subplot(2,1,1)
ax.set_xlabel("Frequency $n$"), ax.set_ylabel("Amplitude $A$")

bx = fig.add_subplot(2,1,2)
bx.set_xlabel("Frequency $n$"), bx.set_ylabel("Phase $\phi$")

ax.stem(x, amplitude, markerfmt="", linefmt="r-", basefmt="r-")
bx.stem(x, phase, markerfmt="", linefmt="b-", basefmt="b-")

bx.yaxis.set_major_locator(MultipleLocator(base=np.pi/4))
bx.yaxis.set_major_formatter(FuncFormatter(pi_formatter))
bx.set_ylim(-(np.pi/2 +0.1), np.pi/2+0.1)

ax.grid(True, linestyle = "--"), bx.grid(True, linestyle = "--")
ax.set_title("Amplitude Spectra"), bx.set_title("Phase Spectra")
fig.suptitle(title, fontweight='bold')
fig.tight_layout()
fig.subplots_adjust(top=0.91)
plt.savefig("spectras-1.png")
plt.show()
```

```python
x_min, x_max = 0, 16

title = f"Espectros de Amplitud y Fase para $g (x)= x^3+x^2$"

x = np.arange(x_min, x_max + 1)
amplitude = (2/(x**3))*np.sqrt((2*x)**2+((np.pi*x)**2-4)**2)
amplitude[0] = np.pi/3
phase = np.arctan((2*x)/(4-(np.pi*x)**2))
    
fig = plt.figure(figsize=(8,8))
ax = fig.add_subplot(2,1,1)
ax.set_xlabel("Frequency $n$"), ax.set_ylabel("Amplitude $A$")

bx = fig.add_subplot(2,1,2)
bx.set_xlabel("Frequency $n$"), bx.set_ylabel("Phase $\phi$")

ax.stem(x, amplitude, markerfmt="", linefmt="r-", basefmt="r-")
bx.stem(x, phase, markerfmt="", linefmt="b-", basefmt="b-")

bx.yaxis.set_major_locator(MultipleLocator(base=np.pi/16))
bx.yaxis.set_major_formatter(FuncFormatter(pi_formatter))
bx.set_ylim(-(np.pi/8 +0.1), np.pi/8+0.1)

ax.grid(True, linestyle = "--"), bx.grid(True, linestyle = "--")
ax.set_title("Amplitude Spectra"), bx.set_title("Phase Spectra")
fig.suptitle(title, fontweight='bold')
fig.tight_layout()
fig.subplots_adjust(top=0.91)
plt.savefig("spectras-2.png")
plt.show()

```
