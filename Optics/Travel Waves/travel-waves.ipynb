```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import os
```

```python
a, b, c = 1, 1, 1
w, k = 1, 1
t=0
for i in ["a", "b", "c", "d", "e", "f"]:
    fig = plt.figure(figsize=(8,4))
    ax = fig.add_subplot(1,1,1)
    if i == "a": 
        x = np.linspace(-5, 4, 1000)
        f = (a*x+b*t+c)**2
        ax.plot(x, f, linewidth = 1, label = f"$\Psi_{i}(x, t=0)$")
    elif i == "b": 
        x = np.linspace(-5, 5, 1000)
        f = np.exp(-(a*x-b*t)**2)
        ax.plot(x, f, linewidth = 1, label = f"$\Psi_{i}(x, t=0)$")
    elif i == "c": 
        x = np.linspace(-np.pi, np.pi, 1000)
        f = 2*np.cos(k*x-w*t)*np.sin(k*x-w*t)
        ax.plot(x, f, linewidth = 1, label = f"$\Psi_{i}(x, t=0)$")
    elif i == "d": 
        x = np.linspace(-5, 5, 1000)
        f = np.tan(a*x+b*t)
        ax.plot(x, f, linewidth = 1, label = f"$\Psi_{i}(x, t=0)$")
    elif i == "e": 
        x = np.linspace(-0.5, 0.5, 100)
        for N in [2, 10, 100]:
            s  = 0
            for n in range(0,N):
                s = s + (x+a*t)**(2*n)
            f = (c*(x+a*t)**2)/(1+np.sqrt(1-(k+1)*(c**2)*(x+a*t)**2)) + s
            ax.plot(x, f, linewidth = 1, label = f"$\Psi_{i}(x, t=0, N = {N})$")
    elif i == "f": 
        x = np.linspace(1, 200, 1000)
        f = np.sin(np.abs(x+a*t)) + np.log(x+a*t)
        ax.plot(x, f, linewidth = 1, label = f"$\Psi_{i}(x, t=0)$")
    
    
    ax.set_title(f"Frente de onda $\Psi_{i}(x,t)$, $[a,b,c = 1; \omega, k = 1, A_n = 1]$")
    ax.legend()
    ax.grid()
    plt.savefig(f"psi_{i}.png")
    plt.show()
```

```python
fig = plt.figure(figsize=(8,4))
ax = fig.add_subplot(1,1,1)
x = np.linspace(0, 2, 1000)
T = np.arange(0,0.06,0.01)
for t in T: 
    E = 50*np.sin(2.5*np.pi*x)*np.cos(20*np.pi*t)
    ax.plot(x, E, linewidth = 1, label = f"$E(x, t={t})$")
    ax.axvline(0, color = "k")
    ax.axvline(2, color = "k")
plt.legend()
plt.grid()
plt.savefig("sw.png")
plt.show()
```

```python
fig = plt.figure(figsize=(8,4))
ax = fig.add_subplot(1,1,1)
x = np.linspace(-5, 5, 1000)
T = np.arange(0,5,1)
for t in T:
    psi = 5*np.sin((2/3)*(x-t))
    ax.plot(x, psi, linewidth = 1, label = f"$E(x, t={t})$")
plt.legend()
plt.grid()
plt.savefig("aw.png")
plt.show()
```
