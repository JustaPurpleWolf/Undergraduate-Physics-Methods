```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import os
```

```python
t = np.linspace(0,10,100)
Z = np.linspace(0,2,2)
fig = plt.figure(figsize=(8,4))
ax = fig.add_subplot(1,1,1)
for z in Z:
    Ex = np.cos(z-t)
    Ey = 2*np.cos(z-t + np.pi/3)
    ax.plot(Ex,Ey)
```
