## Problema 1

El estado más general de polarización se puede expresar de la siguiente manera:
$$\left(\frac{E_x}{A_x}\right)^2 + \left(\frac{E_y}{A_y}\right)^2 - 2 \left(\frac{E_x E_y}{A_x A_y}\right) \cos \phi = \sin^2 \phi,$$
donde $\phi$ es la diferencia de fase $\phi = \phi_y - \phi_x$, $A_x$ y $A_y$ son las amplitudes máximas sobre los ejes $X$ e $Y$ respectivamente. Alternativamente la elipse de polarización se puede escribir
$$\left(\frac{X}{A_x}\right)^2 + \left(\frac{Y}{A_y}\right)^2 - 2 \left(\frac{XY}{A_x A_y}\right) \cos \phi = \sin^2 \phi,$$

a) Mostrar que la envolvente de todas las elipses es un rectángulo de base $A_x$ y altura $A_y$.

b) Hacer una gráfica que muestre la solución del problema como se muestra en la figura para el caso en que $A_x = 2$ y $A_y = 1$, considerando $\phi = 0^\circ, 30^\circ, 60^\circ, 90^\circ, 120^\circ, 150^\circ$ y $180^\circ$.

### Solución

#### Demostración

Veamos que la familia de las elipses de polarización están dadas por $F(X,Y,\phi) = 0$,
$$F(X,Y,\phi) = \left(\frac{X}{A_x}\right)^2 + \left(\frac{Y}{A_y}\right)^2 - 2 \left(\frac{XY}{A_x A_y}\right) \cos \phi - \sin^2 \phi = 0$$
Así, para encontrar la envolvente de ésta familia de elipses derivamos respecto al parámetro $\phi$ e igualamos a cero,
$$\begin{split}
\frac{dF}{d\phi} &= \frac{d}{d\phi}\left[\left(\frac{X}{A_x}\right)^2 + \left(\frac{Y}{A_y}\right)^2 - 2 \left(\frac{XY}{A_x A_y}\right) \cos \phi - \sin^2 \phi\right]\\
&=- 2 \left(\frac{XY}{A_x A_y}\right) \frac{d}{d\phi}[\cos \phi] - \frac{d}{d\phi}[\sin ^2 \phi]\\
&=- 2 \left(\frac{XY}{A_x A_y}\right) (-\sin \phi) - \frac{d}{d\phi}(2 \sin \phi \cos \phi)\\
&= 2 \sin \phi \left[ \frac{XY}{A_x A_y} - \cos \phi \right]\\
&=0
\end{split}$$
Ésto es,
$$\sin \phi \left[ \frac{XY}{A_x A_y} - \cos \phi \right] = 0$$

**1. Caso degenerado** ($\sin \phi = 0$). Ésto ocurre si $\phi = n \pi$, con $n \in \mathbb{Z}$. Como $\phi$ se encuentra en  $[0, 2\pi]$, entonces $\phi = 0, \pi$.
Así, $\sin \phi = 0, \cos \phi = \pm 1$. Sustituyendo en la ecuación de la elipse de polarización obtenemos que,
$$\begin{split}
\left(\frac{X}{A_x}\right)^2 + \left(\frac{Y}{A_y}\right)^2 - 2 \left(\frac{XY}{A_x A_y}\right) \cos \phi = \sin^2 \phi\\
\left(\frac{X}{A_x}\right)^2 + \left(\frac{Y}{A_y}\right)^2 \mp 2 \left(\frac{XY}{A_x A_y}\right) = 0\\
\left(\frac{X}{A_x} \mp \frac{Y}{A_y}\right)^2 = 0\\
\frac{X}{A_x} \mp \frac{Y}{A_y} = 0
\end{split}$$
Así, 
$$Y= \pm \frac{Ay}{A_x}X$$
Ésto son **dos líneas rectas** que pasan por el orígen y van de $-Ax$ a $Ax$ en $X$ (caso degenerado).


**2. Caso no degenerado** ($\frac{XY}{A_x A_y} - \cos \phi = 0$).
Entonces tenemos que, 
$$\begin{split}
\cos \phi &= \frac{XY}{A_x A_y}\\
\sin^2 \phi &= 1 - \cos^2 \phi = 1 - \left(\frac{XY}{A_x A_y}\right)^2 
\end{split}$$
Sustituyendo en la elipse de polarización obtenemos que,
$$\begin{split}
\left(\frac{X}{A_x}\right)^2 + \left(\frac{Y}{A_y}\right)^2 - 2 \left(\frac{XY}{A_x A_y}\right) \cos \phi = \sin^2 \phi\\
\left(\frac{X}{A_x}\right)^2 + \left(\frac{Y}{A_y}\right)^2 - 2 \left(\frac{XY}{A_x A_y}\right)^2 =  1 - \left(\frac{XY}{A_x A_y}\right)^2 \\
\left(\frac{X}{A_x}\right)^2 + \left(\frac{Y}{A_y}\right)^2 -  \left(\frac{XY}{A_x A_y}\right)^2 =  1\\
\left(\frac{Y}{A_y}\right)^2 \left[1 - \left(\frac{X}{A_x}\right)^2 \right] =  \left[1 - \left(\frac{X}{A_x}\right)^2 \right]
\end{split}$$
donde $X \in [-A_x, A_x]$ y $Y \in [-A_y, A_y]$. Veamos que ésto ocurre sólo si, 
$$\left(\frac{Y}{A_y}\right)^2=1 \quad \text{o bien} \quad 1-\left(\frac{X}{A_x}\right)^2 = 0$$

**a)** $\left(\frac{Y}{A_y}\right)^2  = 1$
Entonces, $Y \pm A_y, X \epsilon [-A_x, A_x]$. Obteniendo dos semi-rectas horizontales en $Y \pm A_y$ que van de $-Ax$ a $A_x$.

**b)** $1-\left(\frac{X}{A_x}\right)^2 = 0$
Ésto es,
$$ \left(\frac{X}{A_x}\right)^2=1 $$
Entonces, $X \pm A_x, Y \epsilon [-A_y, A_y]$. Obteniendo dos semi-rectas verticales en $X \pm A_x$ que van de $-Ay$ a $A_y$.

En conjunto podemos ver qué sobre éstos límites, la ecuación forma un rectángulo de base $2 A_X$ y altura $2A_y$ centrado en el origen. 

#### Gráfica
Para una onda polarizada elípticamente con diferencia de fase $\phi$ sabemos que,
$$E_x = A_x \cos (\omega t), \quad E_y = A_y \cos (\omega t + \phi)$$
A partir de éstas ecuaciones podemos construir el espacio fase $(E_x, E_y)$ de las elipses de polarización para $A_x = 2, A_y=1$ a distintas fases $\phi$.

```python
Ax, Ay= 2,1
Phi = [0, 30, 60, 90, 120, 150, 180]
fig = plt.figure(figsize=(8,4))
ax = fig.add_subplot(1,1,1)
# Hacemos w (omega)= 1 debido a que no afecta sobre el diagrama de fase siempre que se cubra un periodo completo de la onda.
t = np.linspace(0, 2*np.pi, 1000) 
for phi in Phi:
    Ex = Ax*np.cos(t)
    Ey = Ay*np.cos(t+phi*np.pi/180)
    ax.plot(Ex,Ey, linewidth = 1, label = f'$\phi={phi}°$')
rect = patches.Rectangle((-Ax,-Ay), 2*Ax, 2*Ay, fill=False, linewidth = 1.5, linestyle = '--', label = "Envolvente")
ax.add_patch(rect)

ax.set_title("Solución Polarización Generalizada", fontweight = 'bold')
ax.set_xlabel(f"$E_x$")
ax.set_ylabel(f"$E_y$")

plt.grid()
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', borderaxespad=0.)
plt.tight_layout()
plt.show()
```

## Problema 2

a) Grafique la reflectancia y transmitancia para una onda con polarizaciones $S$ y $P$ que incide desde el aire sobre un dieléctrico, y encuentre el valor del ángulo de Brewster, considerando que $n_1 = 1$ y $n_2 = 1.5$.

b) Ahora, realice las mismas gráficas para el caso en el que $n_1 = 1.5$ y $n_2 = 1$, encontrando los valores del ángulo de Brewster y del ángulo crítico. Señale la ubicación de dichos ángulos en la gráfica y explique qué es lo que sucede ahí con la transmitancia y la reflectancia.

### Solución

#### Ecuaciones y comportamiento

Recordemos que por la ley de Brewster el **ángulo de Brewster** está dado por,
$$\tan{\theta_B} = \frac{n_2}{n_1}, \quad \theta_B = \arctan\left(\frac{n_2}{n_2}\right)$$
Además sabemos que para $\theta_1 =\theta_B$ tenemos que $R_p  = 0, T_p=1$.

Por otro lado, si $n_1 >n_2$ entonces el **ángulo crítico** $\theta_c$ existe y estará dado por,
$$\sin \theta_c = \frac{n_2}{n_1}, \quad \theta_c = \arcsin\left(\frac{n_2}{n_1}\right)$$ 

Ahora, para la **Reflectancia** $R$ con polarizaciones $S$ (perpendicular) y $P$ (paralela) utilizemos las fórmulas de Fresnel,
$$\begin{split}
R_s = |r_s|^2 ,\quad r_s = \frac{n_1 \cos \theta_1 - n_2 \cos \theta_2}{n_1 \cos \theta_1 + n_2 \cos \theta_2}\\
R_p = |r_p|^2 ,\quad r_s = \frac{n_2 \cos \theta_1 - n_1 \cos \theta_2}{n_2 \cos \theta_1 + n_1 \cos \theta_2}
\end{split}$$
Consideremos que no hay absorción en el material, entonces por conservación de la energía la **Transmitancia** $T$ para cada polarización estará dada por,
$$\begin{split}
T_s = 1- R_s\\
T_p = 1- R_p 
\end{split}$$
Por último para el ángulo de transmisión, por ley de Snell,
$$n_1 \sin \theta_1 = n_2 \sin \theta_2$$
entonces,
$$\theta_2 = \arcsin \left( \frac{n_1}{n_2} \sin \theta_1\right) $$
Veamos que si $n_1 > n_2$ y $\theta_1 \geq \theta_c$ entonces tenemos **Reflexión Total Interna** y $R_s =R_p = 1$, $T_s = T_p=0$.


#### Gráficas

Primero definimos las funciones para el ángulo de transmisión, el ángulo de brewster, el ángulo crítico y los coeficientes de reflexión.

```python
def t2_snell(t1, n1, n2):
    arg = np.clip(n1*np.sin(t1)/n2, -1.0, 1.0) 
    # Usamos numpy.clip para que no se indefina tras pasar el ángulo crítico y veamos el efecto de la refllexión total interna.
    return np.arcsin(arg)
    
def rs(t1, n1, n2):
    t2 = t2_snell(t1,n1,n2)
    return (n1*np.cos(t1)-n2*np.cos(t2))/(n1*np.cos(t1)+n2*np.cos(t2))
    
def rp(t1, n1, n2):
    t2 = t2_snell(t1,n1,n2)
    return (n2*np.cos(t1)-n1*np.cos(t2))/(n2*np.cos(t1)+n1*np.cos(t2))
    
def t_brewster(n1,n2):
    return 180*np.arctan(n2/n1)/np.pi # Multiplicamos por 180/pi paro obtener el ángulo en grados

def t_crit(n1,n2):
    if n1 > n2: return 180*np.arcsin(n2/n1)/np.pi # Verificamos que el ángulo crítico exista.
    else: return None
```

Ahora, definimos una funcón que nos muestre la Reflectancia $R$ y la Transmitancia $T$ sobre el rango del ángulo de incidencia $\theta_1 \epsilon [0, \frac{\pi}{2}]$.

```python
def plot_fresnel(n1,n2):
    fig = plt.figure(figsize=(8,8))
    tb = t_brewster(n1,n2)
    tc = t_crit(n1,n2)
    t1_deg = np.linspace(0, 90, 1000, endpoint=False)
    t1 = t1_deg*np.pi/180 # Convertimos a radianes para utilizar las funciones trigonométricas de numpy.

    ax = fig.add_subplot(2,1,1)
    bx = fig.add_subplot(2,1,2)

    ax.axvline(x = tb,linewidth = 1, linestyle ='--', color = 'gray',label = f'$\\theta_B = {tb:.2f}°$')
    bx.axvline(x = tb,linewidth = 1, linestyle ='--', color = 'gray',label = f'$\\theta_B = {tb:.2f}°$')

    if tc is not None: 
        ax.axvline(x = tc,linewidth = 1, linestyle ='--', color = 'red',label = f'$\\theta_c = {tc:.2f}°$')
        bx.axvline(x = tc,linewidth = 1, linestyle ='--', color = 'red',label = f'$\\theta_c = {tc:.2f}°$')

    Rs, Rp = rs(t1, n1, n2)**2, rp(t1, n1, n2)**2
    ax.plot(t1_deg, Rs, linewidth = 1, label = f'$R_s$')
    ax.plot(t1_deg, Rp, linewidth =1, label = f'$R_p$')

    ax.legend(bbox_to_anchor=(1.05, 1), loc='upper left', borderaxespad=0.)
    ax.set_title("Reflectancia")
    ax.set_xlabel(f"$\\theta_i$ $[deg]$")
    ax.set_ylabel(f"Reflectancia $[abs]$")
    ax.set_xticks(np.arange(0, 100, 10)) 
    ax.grid()

    Ts, Tp = 1-Rs, 1-Rp
    bx.plot(t1_deg, Ts, label = f'$T_s$')
    bx.plot(t1_deg, Tp, label = f'$T_p$')

    bx.legend(bbox_to_anchor=(1.05, 1), loc='upper left', borderaxespad=0.)
    bx.set_title("Transmitancia")
    bx.set_xlabel(f"$\\theta_i$ $[deg]$")
    bx.set_ylabel(f"Transmitancia $[abs]$")
    bx.set_xticks(np.arange(0, 100, 10)) 
    bx.grid()

    plt.suptitle(f"Polarización S y P $(n_1= {n1}, n_2 ={n2})$", fontweight='bold', fontsize=12,y=0.98)
    plt.tight_layout()
    plt.show()
```

Ahora, utilizamos las funciones para cada uno de los casos.

```python
plot_fresnel(n1=1, n2=1.5)
```

```python
plot_fresnel(n1=1.5, n2=1)
```

Por último recordemos que, 
1. Para $\theta_1 =\theta_B$ tenemos que $R_p  = 0, T_p=1$, esto es que el 100% de la luz con polarización $P$ (paralela) se transmite al segundo medio (ángulo de Brewster).
2. Si $n_1 > n_2$ y $\theta_1 \geq \theta_c$ entonces tenemos **Reflexión Total Interna** y $R_s =R_p = 1$, $T_s = T_p=0$ (ángulo crítico).
