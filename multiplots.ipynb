```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def _2graph(x,y1,y2,l,f1,f2,title,name):
    '''
    Genera la gráfica de una función f con formato para su visualización
    x: arreglo de datos del eje horizontal
    y: arreglo de datos del eje vertical
    f: nombre de la función
    title: título de la gráfica
    '''
    # Generar el plot y subplot para la gráfica
    fig = plt.figure(figsize=(10,5))
    ax = fig.add_subplot(1,1,1)

    # Cortar la gráfica para la visualización
    #plt.ylim(l)

    # Colocar los datos sobre la gráfica
    ax.plot(x,y1,linewidth=1.1,label=f1)
    ax.plot(x,y2,linewidth=1.1,label=f2,linestyle="--")

    # Dar formato a la gráfica
    ax.grid()
    ax.set_ylabel('f(x)')
    ax.set_xlabel('x')
    ax.set_title(title)
    ax.legend()

    # Mostrar la gráfica
    plt.savefig(name + '.png')
    plt.show()

def _1graph(x,y,l,f,title,name):
    '''
    Genera la gráfica de una función f con formato para su visualización
    x: arreglo de datos del eje horizontal
    y: arreglo de datos del eje vertical
    f: nombre de la función
    title: título de la gráfica
    '''
    # Generar el plot y subplot para la gráfica
    fig = plt.figure(figsize=(10,5))
    ax = fig.add_subplot(1,1,1)

    # Cortar la gráfica para la visualización
    #plt.ylim(l)

    # Colocar los datos sobre la gráfica
    ax.plot(x,y1,linewidth=1.1,label=f)

    # Dar formato a la gráfica
    ax.grid()
    ax.set_ylabel('f(x)')
    ax.set_xlabel('x')
    ax.set_title(title)
    ax.legend()

    # Mostrar la gráfica
    plt.savefig(name + '.png')
    plt.show()
```

```python
def etominusx(N,x):
    f=(1-np.exp(-np.pi))/np.pi
    for n in range(1,N):
        f+=2/np.pi*(((-1)**(n+1)*np.exp(-np.pi)+1)/(n**2+1))*np.cos(n*x)
    return f
    
termino =50

x = np.linspace(-np.pi, np.pi,5000)
y1 = etominusx(termino,x)
y2 = np.exp(-np.abs(x))
_2graph(x,y1,y2,(-np.pi,-np.pi),'Serie de Fourier','exp(-|x|)', f'Gráfica de la función exp(-|x|) con su serie de Fourier hasta el término N={termino}',f'exp(-x)_N={termino}')
```

```python
def _x(N,x):
    f=0
    for n in range(1,N):
        f+=2*(-1)**(n+1)*np.sin(n*x)/n
    return f
    
termino = 50

x = np.linspace(-np.pi, np.pi,5000)
y1 = _x(termino,x)
y2 = x
_2graph(x,y1,y2,(-np.pi,-np.pi),'Serie de Fourier','x', f'Gráfica de la función x con su serie de Fourier hasta el término N={termino}',f'x_N={termino}')
```

```python
def delta_dirac(N,x):
    f=1/(2*np.pi)
    for n in range(1,N):
        f+=1/np.pi*np.cos(n*x)
    return f
    
termino = 50

x = np.linspace(-np.pi, np.pi,5000)
y1 = delta_dirac(termino,x)
_1graph(x,y1,(-np.pi,-np.pi),'Serie de Fourier', f'Gráfica de la función Delta de Dirac con su serie de Fourier hasta el término N={termino}',f'δ(x)_N={termino}')
```
