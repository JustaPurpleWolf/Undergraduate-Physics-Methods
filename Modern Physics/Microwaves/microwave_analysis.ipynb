```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import os
```

```python
df = pd.read_csv('data/EXP1_wavelength_data_02_03_2026.csv')
d = df['d [cm]']
delta_d = df['delta_d [cm]']
I = df['I [V]']
delta_I = df['delta_I [V]']
```

```python
fig, bx = plt.subplots(1, figsize=(8,4), dpi = 100)
bx.errorbar(d,I,xerr= delta_d, yerr=delta_I)
t = np.linspace(30,35, 1000)
y = np.sin(4*np.pi*t/2.85 - 1.2)
bx.plot(t,y)
plt.show()
```

```python
def plot_char(data_path):
    df = pd.read_csv('data/' + data_path)
    theta = df['theta [deg]']-300
    delta_theta = df['delta_theta [deg]']
    I = df['I [V]']
    delta_I = df['delta_I [V]']
    fig, bx = plt.subplots(1, figsize=(8,4), dpi = 100)
    bx.errorbar(theta,I,xerr= delta_theta, yerr=delta_I)
    plt.show()
```

```python
plot_char('EXP3_cera_data_04_03_2026.csv')
```

```python
plot_char('EXP3_plastilina_data_04_03_2026.csv')
```

```python
plot_char('EXP3_unicel_data_04_03_2026.csv')
```

```python
plot_char('EXP3_madera_data_04_03_2026.csv')
```
