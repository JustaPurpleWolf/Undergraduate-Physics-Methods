```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import os
from scipy.signal import find_peaks
from scipy import constants
from lmfit.models import LinearModel
from lmfit import Model

c = constants.c # Velocidad de la luz [m/s]
e = constants.e # Carga del electrón
```

```python
def plot_wavelength(data_path):
    # 1. Carga de datos
    df = pd.read_csv('data/' + data_path)
    d = df['d [cm]']
    delta_d = df['delta_d [cm]']
    I = df['I [V]']-1
    delta_I = df['delta_I [V]']

    # 2. Detección de picos
    # Ajustamos 'distance' para evitar detectar ruidos como picos
    index, _ = find_peaks(I, height=0.5, distance=5) 
    d_peaks = d.iloc[index].values
    I_peaks = I.iloc[index].values

    # 3. Configuración de la Gráfica
    fig, ax = plt.subplots(figsize= (6, 4), dpi=120)

    # Datos experimentales con barras de error
    ax.errorbar(d, I, xerr=delta_d, yerr=delta_I, 
                fmt='s', markersize=2, capsize=3, 
                ecolor='k', color='k', 
                label='Datos Experimentales', elinewidth = 1.0)
    
    # PARÁMETROS DE DECAIMIENTO (Ajuste Manual Sugerido)
    A0 = I_peaks[0]           # Amplitud inicial en el primer pico
    lamda_ref = 2.84          # Tu valor de lambda anterior
    alpha = 0.11              # COEFICIENTE DE DECAIMIENTO (Ajusta este valor si baja muy rápido)
    d0 = d_peaks[0]           # Punto de referencia para el inicio del decaimiento
    
    # Cálculo de la Curva Teórica con Envolvente
    t = np.linspace(d.min(), d.max(), 1000)
    
    # Envolvente exponencial:
    envolvente = A0 * np.exp(-alpha * (t - d0))
    
    # Función oscilatoria alineada con el primer pico:
    frecuencia = 4 * np.pi / lamda_ref
    # Usamos cos^2 para que el máximo coincida con d0 y no baje de cero
    oscilacion = np.cos(frecuencia / 2 * (t - 30.55))**2
    
    y_teorica = envolvente * oscilacion
    ax.plot(t, y_teorica, color='#1E90FF', linewidth=1, 
            label='Teoría', zorder=2)

    # Resaltado de Máximos Detectados
    ax.scatter(d_peaks, I_peaks, color='red', s=30, marker='v', 
               label='Máximos Detectados', zorder=5)

    # Anotaciones de los picos (opcional, para mayor claridad)
    for i, txt in enumerate(d_peaks):
        ax.annotate(f'{txt:.2f}', (d_peaks[i], I_peaks[i]), 
                    textcoords="offset points", xytext=(0,10), ha='center', 
                    fontsize=7, color='red', fontweight='bold')

    # Estética de Ejes
    ax.set_title(f'Experimento 1. Longitud de onda', 
                 fontsize=14, fontweight='bold', pad=15)
    ax.set_xlabel('Posición $d$ [cm]', fontsize=12)
    ax.set_ylabel('Intensidad $I$ [V]', fontsize=12)
    ax.grid(True, linestyle='--', alpha=0.5)
    ax.legend(loc='upper right')
    ax.set_ylim(-0.15, 2)

    # 4. Guardado y Visualización
    plt.tight_layout()
    plt.savefig('plots/EXP1_wavelength', bbox_inches='tight')
    plt.show()

    # 5. Cálculo de Lambda basado en la diferencia de picos
    if len(d_peaks) > 1:
        diff_d = np.diff(d_peaks)
```

```python
plot_wavelength('EXP1_wavelength_data_02_03_2026.csv')
```

```python
def get_crit(data_path, label):
    df = pd.read_csv('data/' + data_path)
    theta = df['theta [deg]'] - 300
    delta_theta = df['delta_theta [deg]']
    I = df['I [mV]']
    delta_I = df['delta_I [mV]']

    fig, bx = plt.subplots(figsize=(6, 4), dpi=120)
    
    bx.errorbar(theta, I, xerr=delta_theta, yerr=delta_I, 
                fmt='s', markersize=2, capsize=3, ecolor='k', 
                markerfacecolor='k', markeredgecolor='k', elinewidth=0.7,
                label='Datos Experimentales')

    bx.set_title(f'Experimento 3. ' + label, fontsize=14, fontweight='bold')
    bx.set_xlabel('Ángulo $\\theta$ [deg]', fontsize=12)
    bx.set_ylabel('Intensidad $I$ [mV]', fontsize=12)
    bx.grid(True, linestyle='--', alpha=0.7)
    bx.legend()
    
    plt.tight_layout()
    plt.savefig('plots/EXP3_' + label, bbox_inches='tight')
    plt.show()

    beta = theta[np.argmax(I)]
    n = np.sin(np.deg2rad(beta + 30)) / np.sin(np.deg2rad(30))
    
    print("-" * 30)
    print(f"RESULTADOS PARA: {data_path}")
    print(f"β (Máximo)  = {beta:.3f}°")
    print(f"n (Índice)  = {n:.4f}")
    
    if np.abs(n) >= 1:
        theta_crit = np.rad2deg(np.arcsin(1/n))
        print(f"θ꜀ (Crítico) = {theta_crit:.3f}°")
    else:
        print("θ꜀: No definido (n < 1)")
    print("-" * 30)
```

```python
def get_all_crit(path_list, material_list):
    fig, bx = plt.subplots(figsize=(9,7 ), dpi=300)
    
    for data_path, material in zip(path_list, material_list):
        df = pd.read_csv('data/' + data_path)
        theta = df['theta [deg]'] - 300
        delta_theta = df['delta_theta [deg]']
        I = df['I [mV]']
        I0 = I.max()
        delta_I = df['delta_I [mV]']

        beta = theta[np.argmax(I)]
        n = np.sin(np.deg2rad(beta + 30)) / np.sin(np.deg2rad(30))
        if np.abs(n) >= 1:
            theta_crit = np.rad2deg(np.arcsin(1/n))
        else:
            theta_crit = np.nan

        ebar = bx.errorbar(theta, I, xerr=delta_theta, yerr=delta_I, 
                fmt='s', markersize=2, capsize=3, elinewidth=0.7,
                label=f'$\\mathbf{{{material}}}$ (β = {beta:.3f}°, n = {n:.4f}, θc = {theta_crit:.3f}°)')
        color_actual = ebar[0].get_color()
        bx.plot(theta, I, linewidth=0.7, color=color_actual)

    bx.set_title(f'Experimento 3. Caracterización de los Materiales', fontsize=14, fontweight='bold')
    bx.set_xlabel('Ángulo $\\theta$ [deg]', fontsize=12)
    bx.set_ylabel('Intensidad $I$ [mV]', fontsize=12)
    bx.grid(True, linestyle='--', alpha=0.7)
    #bx.legend(bbox_to_anchor=(1.05, 1), loc='upper left', borderaxespad=0.)
    bx.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15),  ncol=2, frameon=False)
    
    plt.tight_layout()
    plt.savefig('plots/EXP3_all.png', bbox_inches='tight')
    plt.show()
```

```python
path_list =['EXP3_cera_data_04_03_2026.csv', 'EXP3_plastilina_data_04_03_2026.csv', 'EXP3_unicel_data_04_03_2026.csv', 
           'EXP3_madera_data_04_03_2026.csv', 'EXP3_agua_data_18_03_2026.csv','EXP3_aceite_data_11_03_2026.csv']
material_list =['Cera', 'Plastilina', 'Unicel', 'Madera', 'Agua', 'Aceite']
get_all_crit(path_list, material_list)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# Modelo físico: Decaimiento exponencial T = A * exp(-2 * kappa * d) + C
def modelo_tunel(d, A, kappa, C):
    return A * np.exp(-2 * kappa * d) + C

def plot_tunnel(data_path, label):
    # 1. Carga de datos
    df = pd.read_csv('data/' + data_path)
    d = df['d [mm]']
    delta_d = df['delta_d [mm]']
    I = df['I [mV]']
    delta_I = df['delta_I [mV]']
    
    # 2. Cálculos de Transmitancia y Propagación de Error
    idx_max = np.argmax(I)
    I0 = I[idx_max]
    delta_I0 = delta_I[idx_max]
    
    T = I / I0
    # Propagación de error para T = I/I0
    delta_T = np.abs(T * np.sqrt((delta_I/I)**2 + (delta_I0/I0)**2))
    
    # 3. Ajuste Exponencial
    try:
        # p0: estimación inicial [Amplitud, kappa, offset]
        popt, pcov = curve_fit(modelo_tunel, d, T, p0=[1, 0.5, 0])
        d_fit = np.linspace(d.min(), d.max(), 500)
        T_fit = modelo_tunel(d_fit, *popt)
        error_kappa = np.sqrt(pcov[1,1]) # Incertidumbre del ajuste
    except:
        T_fit = None

    # 4. Configuración de la Gráfica
    fig, ax = plt.subplots(figsize=(6, 4), dpi=120)
    
    # Datos experimentales (Limpio, sin saturación)
    ax.errorbar(d, T, xerr=delta_d, yerr=delta_T, 
                fmt='s', markersize=2, capsize=3, 
                ecolor='black', markerfacecolor='black', 
                markeredgecolor='black', label='Datos Experimentales (T)',
                alpha=0.8, elinewidth=0.7)

    # Línea de ajuste (Modelo Teórico)
    if T_fit is not None:
        ax.plot(d_fit, T_fit, color='#D2691E', linewidth=1, 
                label=f'Ajuste Exponencial ($\kappa$={popt[1]:.3f})')

    # Estética y Etiquetas
    ax.set_title(f'Experimento 4. ' + label, fontsize=14, fontweight='bold')
    ax.set_xlabel('Distancia $d$ [mm]', fontsize=12)
    ax.set_ylabel('Transmitancia $T$ ($I/I_0$)', fontsize=12)
    ax.grid(True, which="both", linestyle='--', alpha=0.6)
    ax.legend()

    # 5. Guardado y Cierre
    plt.tight_layout()
    plt.savefig('plots/EXP4_'+ label, bbox_inches='tight')
    plt.show()

    # 6. Reporte de Resultados
    print("-" * 40)
    print(f"RESULTADOS EFECTO TÚNEL: {data_path}")
    if T_fit is not None:
        print(f"Parámetro de decaimiento (κ): {popt[1]:.4f} ± {error_kappa:.4f} mm⁻¹")
        print(f"Amplitud inicial (A): {popt[0]:.4f}")
    print("-" * 40)
```

```python
def plot_malus(data_path):
    # 1. Carga y preparación de datos
    df = pd.read_csv('data/' + data_path)
    
    # Ajuste de fase (Normalización de ángulos a partir de 0)
    theta = df['theta [deg]'] - 180
    delta_theta = df['delta_theta [deg]']
    I = df['I [mV]']
    delta_I = df['delta_I [mV]']
    
    # Normalización de Intensidad
    I0 = np.max(I)
    I_norm = I / I0
    delta_I_norm = delta_I / I0

    # 2. Configuración de la Gráfica
    fig, ax = plt.subplots(figsize=(6, 4), dpi=120)

    # Datos experimentales
    ax.errorbar(theta, I_norm, xerr=delta_theta, yerr=delta_I_norm, 
                fmt='s', markersize=5, capsize=3, 
                ecolor='k', markerfacecolor='k', 
                markeredgecolor='k', label='Datos Experimentales ($I/I_{max}$)',
                alpha=0.8, zorder=3)

    # Curva Teórica (Ley de Malus)
    t = np.linspace(0, 360, 1000)
    y_teorica = np.cos(np.deg2rad(t))**2
    ax.plot(t, y_teorica, color='#FF4500', linewidth=1, 
            label='Teoría ($\cos^2(\\theta)$)', zorder=2, alpha=0.9)

    # Personalización de Ejes y Estética
    ax.set_title(f'Experimento 2. Ley de Malus', fontsize=14, fontweight='bold', pad=15)
    ax.set_xlabel('Ángulo relativo $\\theta$ [deg]', fontsize=12)
    ax.set_ylabel('Intensidad Normalizada $I/I_0$', fontsize=12)
    
    # Forzar ticks cada 45 grados para mejor lectura
    ax.set_xticks(np.arange(0, 361, 45))
    ax.grid(True, linestyle=':', alpha=0.6)
    ax.set_ylim(-0.05, 1.1) # Margen para ver bien los máximos y mínimos
    ax.legend(loc='upper right', frameon=True, shadow=True)

    # 3. Guardado y Visualización
    plt.tight_layout()
    plt.savefig('plots/EXP2_malus', bbox_inches='tight')
    plt.show()

    # 4. Reporte rápido
    print("-" * 45)
    print(f"ANÁLISIS DE POLARIZACIÓN: {data_path}")
    print(f"Intensidad Máxima detectada (I0): {I0:.2f} mV")
    print(f"Rango angular cubierto: {theta.min()}° a {theta.max()}°")
    print("-" * 45)
```

```python
plot_malus('EXP2_malus_data_18_03_2026.csv')
```

```python
def abs_lineal(x, amplitude, center, intercept):
    """Modelo de valor absoluto para una línea."""
    return amplitude * np.abs(x - center) + intercept

def plot_malus_2(data_path):
    # 1. Carga y preparación de datos
    df = pd.read_csv('data/' + data_path)
    
    # Ajuste de fase (Normalización de ángulos a partir de 0)
    theta = df['theta [deg]'] - 180
    delta_theta = df['delta_theta [deg]']
    I = df['I [mV]']
    delta_I = df['delta_I [mV]']
    
    # Normalización de Intensidad
    I0 = np.max(I)
    I_ln_0 = np.log(I / I0)
    dI_ln_0 = delta_I / I

    cos_t = np.log(np.abs(np.cos(np.deg2rad(theta))))
    dcos_t = np.abs(np.tan(np.deg2rad(theta))*np.deg2rad(delta_theta))

    mask = cos_t >= -1
    ln_cos = cos_t[mask]
    dln_cos = dcos_t[mask]
    I_ln = I_ln_0[mask]
    dI_ln = dI_ln_0[mask]

    # Ajuste lineal
    model = LinearModel()
    params = model.guess(I_ln, x=ln_cos)
    result = model.fit(I_ln, params, x=ln_cos)
    print(result.fit_report())
    result.plot()
    plt.show()

    
    # 2. Configuración de la Gráfica
    fig, ax = plt.subplots(figsize=(6, 4), dpi=120)
    
    # Datos experimentales
    ax.errorbar(ln_cos, I_ln, xerr=dln_cos, yerr=dI_ln, 
                fmt='s', markersize=5, capsize=3, 
                ecolor='k', markerfacecolor='k', 
                markeredgecolor='k', label='Datos Experimentales',
                alpha=0.8, zorder=3)
    
    # Gráfica teórica
    pendiente = result.params['slope'].value
    error_pendiente = result.params['slope'].stderr

    ordenada = result.params['intercept'].value
    error_ordenada = result.params['intercept'].stderr

    print(f"El grado de relación es n = {pendiente:.4f} ± {error_pendiente:.4f}")

    x_grafica = np.linspace(-1, 0, 100) 

    # Reconstruir la recta: y = m*x + b
    y_recuperada = pendiente * x_grafica + ordenada

    plt.plot(x_grafica, y_recuperada, 'r--', label=f'n= {pendiente:.2f}', linewidth = 1.0)
    
    # Personalización de Ejes y Estética
    ax.set_title(f'Experimento 2. Linearidad del Receptor', fontsize=14, fontweight='bold', pad=15)
    ax.set_xlabel('Intensidad relativa $ln(I/I_0) = ln(cos \\theta))$', fontsize=12)
    ax.set_ylabel('Respuesta del receptor $ln(V/V_0)$', fontsize=12)
    
    ax.grid(True, linestyle=':', alpha=0.6)
    ax.legend(loc='lower right', frameon=True, shadow=True)

    # 3. Guardado y Visualización
    plt.tight_layout()
    plt.savefig('plots/EXP2_malus_2', bbox_inches='tight')
    plt.show()
```

```python
plot_malus_2('EXP2_malus_data_18_03_2026.csv')
```

```python
def plot_all_tunnel(path_list, material_list):
    fig, ax = plt.subplots(figsize=(9,7 ), dpi=300)
    
    for data_path, material in zip(path_list, material_list):
        # 1. Carga de datos
        df = pd.read_csv('data/' + data_path)
        d = df['d [mm]']
        delta_d = df['delta_d [mm]']
        I = df['I [mV]']
        delta_I = df['delta_I [mV]']
    
        # 2. Cálculos de Transmitancia y Propagación de Error
        idx_max = np.argmax(I)
        I0 = I[idx_max]
        delta_I0 = delta_I[idx_max]
    
        T = I / I0
        # Propagación de error para T = I/I0
        delta_T = np.abs(T * np.sqrt((delta_I/I)**2 + (delta_I0/I0)**2))
    
        # 3. Ajuste Exponencial
        try:
            # p0: estimación inicial [Amplitud, kappa, offset]
            popt, pcov = curve_fit(modelo_tunel, d, T, p0=[1, 0.5, 0])
            d_fit = np.linspace(d.min(), d.max(), 500)
            T_fit = modelo_tunel(d_fit, *popt)
            error_kappa = np.sqrt(pcov[1,1]) # Incertidumbre del ajuste
        except:
            T_fit = None
            popt[1] = np.nan
    
        # Datos experimentales (Limpio, sin saturación)
        ebar=ax.errorbar(d, T, xerr=delta_d, yerr=delta_T, fmt='s', markersize=2, capsize=3, 
                         label=f'$\\mathbf{{{material}}}$ ($\kappa$={popt[1]:.3f})', alpha=0.8, elinewidth=0.7)
        color_actual = ebar[0].get_color()
        # Línea de ajuste (Modelo Teórico)
        if T_fit is not None:
            ax.plot(d_fit, T_fit, color=color_actual, linewidth=1)

    # Estética y Etiquetas
    ax.set_title(f'Experimento 4. Tunelamiento de la Onda Evanescente', fontsize=14, fontweight='bold')
    ax.set_xlabel('Distancia $d$ [mm]', fontsize=12)
    ax.set_ylabel('Transmitancia $T$ ($I/I_0$)', fontsize=12)
    ax.grid(True, which="both", linestyle='--', alpha=0.6)
    ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15),  ncol=2, frameon=False)

    # 5. Guardado y Cierre
    plt.tight_layout()
    plt.savefig('plots/EXP4_all', bbox_inches='tight')
    plt.show()
```
