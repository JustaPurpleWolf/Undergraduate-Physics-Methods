```python
dataset = { # 'nombre_del_archivo': cantidad_de_hojas
    '0001002A': 500,
    '0001003A': 450,
    '0001005A': 407,
    '0001006A': 365,
    '0001007A': 341,
    '0001008A': 297,
    '0001009A': 254,
    '0001010A': 221,
    '0001011A': 186,
    '0001012A': 141,
    '0001013A': 107,
    '0001014A': 97,
    '0001015A': 87,
    '0001016A': 77,
    '0001017A': 72,
    '0001018A': 62,
    '0001019A': 42,
    '0001020A': 52,
    '0001021A': 32,
    '0001022A': 12,
    '0001023A': 2,
    '0001024A': 1,
    '0001025A': 5,
}
unit_thickness = 0.015 # mm por hoja
historial = []
```

```python
for file, count in dataset.items():
    path = os.path.join('DICOM', file)
    if os.path.exists(path):
        try:
            rad = dicom.load(path)
            paper = rad.invert().crop(550, 600, 1300, 1750)
            paper_thickness = count * unit_thickness
            _, _, info = paper.measure_weighted(
                label="Serie_Calibracion", 
                thickness=paper_thickness, 
                sheets=count
            )
            historial.append(info)
            print(f"{path} procesado con éxito.")
            
        except Exception as e:
            print(f"Error procesando {path}; {e}")
    else:
        print(f"Archivo {path} no encontrado.")
```

```python
dicom.save_to_csv(historial, overwrite = True)
```

```python
plot_calibration('calibration_thickness.csv', unit_thickness, mu=0.032) # mm por hoja
```
