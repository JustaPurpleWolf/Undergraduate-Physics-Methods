```python
import matplotlib.pyplot as plt
import numpy as np

# ==========================================
# CONFIGURACIÓN DE ESTILO PARA ARTÍCULO
# ==========================================
plt.rcParams.update({
    "font.family": "serif",
    "font.serif": ["Times New Roman"],
    "axes.labelsize": 10,
    "xtick.labelsize": 9,
    "ytick.labelsize": 9,
    "legend.fontsize": 8.5,
    "figure.dpi": 300,
})

# ==========================================
# PARÁMETROS DEL SISTEMA ATÓMICO
# ==========================================
Gamma_31 = 6.0   # Tasa de decaimiento (MHz)
Gamma_21 = 0.02  # Tasa de decoherencia (MHz)
k0 = 1.0         # Número de onda normalizado
L = 2.5          # Longitud de la celda

Delta_p = np.linspace(-15, 15, 1200)
Omega_c_values = [2.0, 5.0, 10.0]
Omega_c_demo = 5.0
delta = Delta_p

# ==========================================
# CREACIÓN DE LA FIGURA (3 SUBPLOTS SIN TÍTULOS)
# ==========================================
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(14, 3.6))

# ------------------------------------------
# 1. Susceptibilidad Eléctrica (Re y Im)
# ------------------------------------------
num_chi = 1j * (Gamma_21 - 1j * delta)
den_chi = (Gamma_31 - 1j * Delta_p) * (Gamma_21 - 1j * delta) + (Omega_c_demo**2) / 4.0
chi = num_chi / den_chi

ax1.plot(Delta_p, np.real(chi), label="$\\text{Re}[\\chi]$", color="black", linestyle="--", linewidth=1.2)
ax1.plot(Delta_p, np.imag(chi), label="$\\text{Im}[\\chi]$", color="black", linestyle="-", linewidth=1.2)
ax1.set_xlabel("Desintonía $\\Delta_p$ (MHz)")
ax1.set_ylabel("Susceptibilidad $\\chi$ [U.A.]")
ax1.axhline(0, color="gray", linestyle=":", linewidth=0.6)
ax1.axvline(0, color="gray", linestyle=":", linewidth=0.6)
ax1.grid(False)
ax1.legend(frameon=False)

# ------------------------------------------
# 2. Coeficiente de Absorción y Ensanchamiento
# ------------------------------------------
styles = ["-", "--", "-."]
for i, Omega_c in enumerate(Omega_c_values):
    num_c = Gamma_31 * ((Gamma_21**2) * Gamma_31 + Gamma_21 * (Omega_c**2 / 4.0) + Gamma_21 * (Delta_p**2))
    den_c = (Gamma_31 * Gamma_21 + (Omega_c**2 / 4.0) - Delta_p**2)**2 + (Delta_p**2 * (Gamma_31 + Gamma_21)**2)
    alpha = k0 * (num_c / den_c)
    
    ax2.plot(Delta_p, alpha, label=f"$\\Omega_c = {Omega_c}$ MHz", color="black", linestyle=styles[i], linewidth=1.2)

ax2.set_xlabel("Desintonía $\\Delta_p$ (MHz)")
ax2.set_ylabel("Coeficiente de absorción $\\alpha$ [cm$^{-1}$]")
ax2.axhline(0, color="gray", linestyle=":", linewidth=0.6)
ax2.axvline(0, color="gray", linestyle=":", linewidth=0.6)
ax2.grid(False)
ax2.legend(frameon=False)

# ------------------------------------------
# 3. Transmisión Óptica (Ley de Beer-Lambert)
# ------------------------------------------
for i, Omega_c in enumerate(Omega_c_values):
    num_c = Gamma_31 * ((Gamma_21**2) * Gamma_31 + Gamma_21 * (Omega_c**2 / 4.0) + Gamma_21 * (Delta_p**2))
    den_c = (Gamma_31 * Gamma_21 + (Omega_c**2 / 4.0) - Delta_p**2)**2 + (Delta_p**2 * (Gamma_31 + Gamma_21)**2)
    alpha = k0 * (num_c / den_c)
    T = np.exp(-alpha * L)
    
    ax3.plot(Delta_p, T, label=f"$\\Omega_c = {Omega_c}$ MHz", color="black", linestyle=styles[i], linewidth=1.2)

ax3.set_xlabel("Desintonía $\\Delta_p$ (MHz)")
ax3.set_ylabel("Transmisión $\\mathscr{T}$")
ax3.axvline(0, color="gray", linestyle=":", linewidth=0.6)
ax3.grid(False)
ax3.legend(frameon=False)

# ==========================================
# EXPORTACIÓN Y VISUALIZACIÓN
# ==========================================
plt.tight_layout()

plt.show()
```
