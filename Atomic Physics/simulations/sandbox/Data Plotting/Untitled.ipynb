```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import os
from lmfit import Model
```

```python
label = 'F0009' 
df1 = pd.read_csv(label+'CH1.csv',header=None)
df2 = pd.read_csv(label+'CH2.csv',header=None)
t1, t2 = df1[3], df2[3]
I1, I2 = df1[4]/50, df2[4]/500
```

```python
fig, ax = plt.subplots(figsize=(6, 4), dpi=150)
ax.set_xlabel(r'time [$s$]', fontweight='bold')
ax.set_ylabel(r'Intensity [$mV$]', fontweight='bold')
ax.plot(t1,I1, linewidth = 0.6,color="blue", label = 'spec', zorder =2)
ax.plot(t2,I2, linewidth = 0.6,color="red", label = 'satspec', zorder =2)

ax.grid(True, which='both', alpha=0.5)
ax.legend(loc='lower right', frameon=True)

ax.set_title('Spectroscopy in RbD2 line at T=73.5°C', pad=10, fontweight = 'bold')
plt.tight_layout()
plt.savefig('image_plot_'+label+'.png', bbox_inches='tight')
plt.show()
```
